# **Mensagens-chave das aulas**

**Mensagem-chave: Dados organizados**: os dados precisam estar organizados para facilitar e otimizar o processo de análise, feito pela Máquina. Portanto, os dados precisam estar organizados no padrão da Máquina. O padrão de organização (tyde data) da máquina consiste em três regras: cada tipo de informação deve estar em uma coluna diferente; cada observação/cenário deve estar em uma linha diferente, e os dados ficam nas células da matriz/tabela.

**Mensagem-chave: recursos Python para manipulação de dados tabulares**: os dados tabulares precisam ser manupulados para viabilizar a conversão para uma estrutura organizada. Os tipos de manipulação vistos são: preenchimento de valores/células vazias, padronizar rótulo das colunas (variáveis), ordenação de valores, eliminação de duplicadas, remoção de espaçõs e pádronização das informações, crinar novas colunas/variáveis com valores calculados, filtragem por colunas e valores.

**Mensagem-chave: reestruturação e pivoteamento de dados tabulares:** a reestruturação e pivoteamento de dados tabulares são realizadas com a finalidade de modificar a estrutura da tabela, para facilitar a análise por máquinas ou humanos. A estrutura das tabelas pode estar no formato largo (humanos), onde as informações podem aumentar tanto na horizontal (colunas)como na vertical (linhas). A estrutura pode estar também no formato longo (máquina - dados organizados), onde as informações aumentam somente na vertical (linhas). As funções da biblioteca Pandas utilizadas para reestruturação e pivotamento são: melt() e pivot(), respectivamente.

**Mensagem-chave: Fusão e Combinação de dados:** a fusão e combinação de dados é importante para juntar dados disponíveis/disponibilizados em diferentes locais/bases, enriquecendo uma base de dados comum com todas as informações. Os dados fusionados/concatenados já ocorrem na forma organizada (máquina-longo). A concatenação é utilizada para juntar bases de dados equivalentes, ou seja, com linhas/colunas em comum. A concatenação pode ser realizada na vertical (empilhar), juntando os dados por colunas a aumentando o número de linhas (axis=0), ou na horizontal, juntando os dados por linhas e aumentando o número de colunas (axis=1). Na biblioteca Pandas a função concat() é utilizada para executar a junção. Já a fusão de dados, é utilizada para juntar diferentes bases de dados com informações complementares entre si. A fusão ocorre com base em uma coluna "âncora" (que é comum as duas bases: CPF, id_produto, ...). Um cuidado a ser tomado na operação de fusão da base de dados é em relação ao modo da fusão (que pode ser inner, outer, left, right). O modo "inner" (intersecção entre as duas bases) pode resultar em uma perca significativa de informações, caso existam muitos valores que não estão presentes nas duas bases. No pandas, a função utilizada para a operação de fusão é a merge(), equivalente ao JOIN (sql) ou PROCV (excel). Vale ressantar que, para ser possível a fusão entre bases de dados, é necessário que exista uma coluna/variável em comum entre as bases.

**Mensagem-chave: Indexação e consultas de subconjuntos da tabela**: A finalidade de um índice é definir coordenadas do endereço das informações dentro de uma tabela. Com as coordenadas/endereço de cada informações na uma tabela, conseguimos utilizar mecanismos de busca para filtrar subconjuntos (valor isolado, linha, coluna, matriz) de interesse que estão contidos na tabela. Sendo assim, os índices servem para especificar o endereço das informações dentro de uma matriz. Existem dois tipos de índice: o padrão (número das linhas e colunas - começa contar em 0), e o índice semântico (valores/labels de uma coluna definida e os rótulos das colunas). Um ponto negativo do índice padrão é que, em casos de reordenação das linhas/colunas, um mesmo valor pode mudar de índice (endereço). Para fazer as consultas por meio dos índices utilizamos as funções loc - para índices semânticos/labes e iloc - para índices padrões/números.

## Atividade 1

```python
import pandas as pd
import numpy as np
```

```python
# 1. Carragar DataFrame
dados = {
    'ID Vendedor': [101, 102, 101, 103, 104],
    'Região': ['Sul', 'NORTE', 'Sul', 'sul', 'Leste'],
    'FATURAMENTO': [5000, 3000, 5000, np.nan, 4500],
    'Email': [
        'ana@furb.br',
        'bruno@furb.br',
        'ana@furb.br',
        'carla@furb.br',
        'joao@furb.br',
    ]
}

df = pd.DataFrame(dados)
print(df)

# Carregar diretamento do Excel
#df = pd.read_excel("file_name")
```

```python
# 2. Renomeação das colunas

df1 = df.rename(columns={
    'ID Vendedor': 'ID_VENDEDOR',
    'Região': 'REGIAO',
    'Email': 'EMAIL'
})

print(df1)
```

```python
# 3. Eliminação de duplicatas

df2 = df1.drop_duplicates(subset=['ID_VENDEDOR'])
print(df2)
```

```python
# 4. Tratamento de Dados Ausentes

df2['FATURAMENTO'] = df2['FATURAMENTO'].fillna(df2['FATURAMENTO'].mean())

print(df2)
```

```python
# 5. Criação de uma coluna calculada

df2['FATURAMENTO-ANUAL'] = df2['FATURAMENTO']*12
print(df2)
```

```python
# 5. Criação de uma nova coluna com lista de valores

df2['PERC-COMISSAO'] = [10, 20, 15, 10]
print(df2)
```

```python
# 5. Criação de uma nova coluna operando com mais de uma coluna

df2['LUCRO'] = df2['FATURAMENTO-ANUAL'] - df2['PERC-COMISSAO']/100 * df2['FATURAMENTO-ANUAL']
print(df2)
```

```python
# 6. Padronização e ordenação

df2['REGIAO'] = df2['REGIAO'].str.lower()
df2 = df2.sort_values(by='FATURAMENTO', ascending=False)

print(df2)
```

```python
# 7. Exportação Final

df2.to_excel('dados_processados.xlsx', index=False)
df2.to_csv('dados_processados.csv', index=False)
```

## Atividade 2

```python
import pandas as pd
```

```python
# 1. Carregando os dados no Formato Largo (Wide)
df_notas = pd.DataFrame({
    'Estudante': ['Alice', 'Bruno', 'Carla'],
    'Matematica': [8.5, 7.0, 9.5],
    'Ciencias': [9.0, 6.5, 8.0],
    'Historia': [7.0, 8.0, 8.5]
})

print(df_notas)
```

```python
# 2. Transformando em Formato Longo com pd.melt()

df_longo = pd.melt(df_notas,
                   id_vars=['Estudante'],
                   value_vars=['Matematica', 'Ciencias', 'Historia'],
                   var_name='Disciplina',
                   value_name='Nota')

print(df_longo)
```

```python
# 3. Reorganizando com pd.pivot()

df_boletim = df_longo.pivot(index='Estudante',
                            columns='Disciplina',
                            values='Nota')

print("\nTabela Pivoteada (Retorno ao Formato Largo):")
print(df_boletim)
```

## Atividade 3

```python
import pandas as pd
```

```python
# 1. Criando as tabelas

df_alunos = pd.DataFrame({
    'id_aluno': [1, 2, 3, 4, 5],
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Diana', 'Eduardo']
})

df_novos_alunos = pd.DataFrame({
    'id_aluno': [6, 7, 8, 9, 10],
    'Nome': ['Fernanda', 'Gabriel', 'Helena', 'Igor', 'Julia']
})

df_notas = pd.DataFrame({
    'id_aluno': [1, 2, 4, 6, 8, 11],
    'Nota_Final': [8.5, 7.0, 9.2, 6.5, 8.0, 10.0]
})

print('df_alunos')
print(df_alunos,'\n')
print('df_novos_alunos')
print(df_novos_alunos,'\n')
print('df_notas')
print(df_notas)
```

```python
# 2. Empilhar as listas de alunos

df_alunos_completo = pd.concat([df_alunos, df_novos_alunos], ignore_index=True)

print('df_alunos_completo')
print(df_alunos_completo)
```

```python
# 3. Cruzamento da lista de alunos com o registro das notas

df_relatorio_final = pd.merge(df_alunos_completo,
                              df_notas,
                              on='id_aluno',
                              how='outer')

print("df_relatorio_final")
print(df_relatorio_final)
```

## Atividade prática 1

```python
import pandas as pd
import numpy as np

# Dataframe com os dados
df = pd.DataFrame({
    'Nome': ['Alberto', 'Bárbara', 'Cláudia', 'Pedro', 'Rosa', 'Rui'],
    'Batimento': [65, 88, 74, np.nan, np.nan, 60],
    'Pressão': [132, 90, np.nan, 115, 86, 138],
    'Temperatura': [38, np.nan, 38, 36, 37, 39],
    'Sintoma': ['Nenhum', np.nan, np.nan, 'Dor no peito', np.nan, np.nan],
    'Diagnóstico': ['Doente', 'Saudável', 'Saudável', 'Doente', 'Saudável', 'Doente']
})

# Exibindo a tabela
print(df)
```

```python
# Renomeação das colunas
df = df.rename(columns={
    'Nome': 'NOME',
    'Batimento': 'BATIMENTO',
    'Pressão': 'PRESSAO',
    'Temperatura': 'TEMPERATURA',
    'Sintoma': 'SINTOMA',
    'Diagnóstico': 'DIAGNOSTICO'
})


# Padronização textos
df['SINTOMA'] = df['SINTOMA'].str.strip().str.lower()
df['DIAGNOSTICO'] = df['DIAGNOSTICO'].str.strip().str.lower()

print(df)
```

```python
# Criar uma cópia independente do DataFrame original
df2 = df.copy()
```

```python
# Criando o indicador de ausência (1 se for NaN, 0 se tiver valor)
# 'isna()': retorna True para valores ausentes e False para valores preenchidos
# 'astype(int)': converte True em 1 e False em 0
batimento_ausente = df2['BATIMENTO'].isna().astype(int)

# Inserindo a nova coluna na sequência (índice da coluna + 1)
df2.insert(loc=df2.columns.get_loc("BATIMENTO") + 1,
          column='BATIMENTO_AUS',
          value=batimento_ausente)

print(df2)
```

```python
# Preencher valores ausentes da coluna "Batimento"
df2["BATIMENTO"] = df2["BATIMENTO"].fillna(round(df2["BATIMENTO"].mean()))

print(df2)
```

```python
df3 = df2.copy()

# Tratamento da coluna Sintomas
sintoma_ausente = df3['SINTOMA'].isna().astype(int)

# Inserindo a nova coluna na sequência (índice da coluna + 1)
df3.insert(loc=df3.columns.get_loc("SINTOMA") + 1,
          column='SINTOMA_AUS',
          value=sintoma_ausente)

# Preencher valores ausentes
df3["SINTOMA"] = df3["SINTOMA"].fillna("sem sintoma")
print(df3)
```

### Atividade prática 2

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'nome': ['Ana', 'Bruno', 'Carlos', 'Daniel', 'Ana', 'Elena', 'Fabio', 'Gisele'],
    'idade': [21, 176, 1, 26, 21, 41, 34, 28],
    'data_nascimento': ['1999-05-10', '1844-01-01', '2019-03-20', '1994-08-15', '1999-05-10', '1979-11-30', '1986-02-12', '1992-07-05'],
    'temperatura': [36.5, 37.0, -5.0, 36.8, 36.5, 41.5, 37.2, 36.6],
    'frequencia_cardiaca': [75, 80, 72, np.nan, 75, 110, 999, 82],
    'escolaridade': ['Graduação', 'Mestrado', 'Doutorado', 'Ensino Médio', 'Graduação', 'Graduação', 'Ensino Médio', 'Pós-graduação']
})

print(df)
```

```python
# Identifica e remove registros idênticos
print(df[df.duplicated(keep=False)])
df2 = df.drop_duplicates(keep='last')
print('\n', df2)
```

```python
df2 = df2.drop(columns=['data_nascimento'])
print(df2)
```

```python
# Tratamento de dados ausentes
# Contar quantos nulos existem por coluna
print("\nDiagnóstico de NaNs:\n", df2.isnull().sum(), '\n')
```

```python
# Tratamento de dados inconsistentes e outliers
# Criando o indicador de ausência
frequencia_ausente = df2['frequencia_cardiaca'].isna().astype(int)
df2.insert(loc=df2.columns.get_loc("frequencia_cardiaca") + 1,
          column='FREQ_CARDIACA_AUS',
          value=frequencia_ausente)

# Preencher a frequência cardíaca vazia com a média
mediana_frequencia = round(df2['frequencia_cardiaca'].median())
df2['frequencia_cardiaca'] = df2['frequencia_cardiaca'].fillna(mediana_frequencia)

print(df2)
```

```python
df3 = df2.copy()
```

```python
# CORREÇÕES DE INCONSISTÊNCIAS E RUÍDOS

# Criando o indicador de dado corrigido
idade_corrigida = (df3['idade'] > 120).astype(int)
df3.insert(loc=df3.columns.get_loc("idade") + 1,
          column='IDADE_COR',
          value=idade_corrigida)

# Identificar e tratar idade impossível
media_idade = round(df3.loc[df3['idade'] <= 120, 'idade'].mean())
df3.loc[df3['idade'] > 120, 'idade'] = media_idade


# Identificar e tratar temperatura impossível (ex: < 30°C)
media_temp = round(df3.loc[df3['temperatura'] >= 30, 'temperatura'].mean(), 1)
df3.loc[df3['temperatura'] < 30, 'temperatura'] = media_temp


# Identificar e tratar ruído na frequência cardíaca
media_frequencia = round(df3.loc[df['frequencia_cardiaca'] <= 300, 'frequencia_cardiaca'].mean())
df3.loc[df['frequencia_cardiaca'] == 999, 'frequencia_cardiaca'] = media_frequencia

print(df3)
```
