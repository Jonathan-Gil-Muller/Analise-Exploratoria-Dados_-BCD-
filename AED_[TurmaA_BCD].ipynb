# **Mensagens-chave das aulas**

**AULA 1 Mensagem-chave: Dados organizados**: os dados precisam estar organizados para facilitar e otimizar o processo de análise, feito pela Máquina. Portanto, os dados precisam estar organizados no padrão da Máquina. O padrão de organização (tyde data) da máquina consiste em três regras: cada tipo de informação deve estar em uma coluna diferente; cada observação/cenário deve estar em uma linha diferente, e os dados ficam nas células da matriz/tabela.

**AULA 1 Mensagem-chave: recursos Python para manipulação de dados tabulares**: os dados tabulares precisam ser manupulados para viabilizar a conversão para uma estrutura organizada. Os tipos de manipulação vistos são: preenchimento de valores/células vazias, padronizar rótulo das colunas (variáveis), ordenação de valores, eliminação de duplicadas, remoção de espaçõs e pádronização das informações, crinar novas colunas/variáveis com valores calculados, filtragem por colunas e valores.

**AULA 2 Mensagem-chave: reestruturação e pivoteamento de dados tabulares:** a reestruturação e pivoteamento de dados tabulares são realizadas com a finalidade de modificar a estrutura da tabela, para facilitar a análise por máquinas ou humanos. A estrutura das tabelas pode estar no formato largo (humanos), onde as informações podem aumentar tanto na horizontal (colunas)como na vertical (linhas). A estrutura pode estar também no formato longo (máquina - dados organizados), onde as informações aumentam somente na vertical (linhas). As funções da biblioteca Pandas utilizadas para reestruturação e pivotamento são: melt() e pivot(), respectivamente.

**AULA 2 Mensagem-chave: Fusão e Combinação de dados:** a fusão e combinação de dados é importante para juntar dados disponíveis/disponibilizados em diferentes locais/bases, enriquecendo uma base de dados comum com todas as informações. Os dados fusionados/concatenados já ocorrem na forma organizada (máquina-longo). A concatenação é utilizada para juntar bases de dados equivalentes, ou seja, com linhas/colunas em comum. A concatenação pode ser realizada na vertical (empilhar), juntando os dados por colunas a aumentando o número de linhas (axis=0), ou na horizontal, juntando os dados por linhas e aumentando o número de colunas (axis=1). Na biblioteca Pandas a função concat() é utilizada para executar a junção. Já a fusão de dados, é utilizada para juntar diferentes bases de dados com informações complementares entre si. A fusão ocorre com base em uma coluna "âncora" (que é comum as duas bases: CPF, id_produto, ...). Um cuidado a ser tomado na operação de fusão da base de dados é em relação ao modo da fusão (que pode ser inner, outer, left, right). O modo "inner" (intersecção entre as duas bases) pode resultar em uma perca significativa de informações, caso existam muitos valores que não estão presentes nas duas bases. No pandas, a função utilizada para a operação de fusão é a merge(), equivalente ao JOIN (sql) ou PROCV (excel). Vale ressantar que, para ser possível a fusão entre bases de dados, é necessário que exista uma coluna/variável em comum entre as bases.

**AULA 2 Mensagem-chave: Indexação e consultas de subconjuntos da tabela**: A finalidade de um índice é definir coordenadas do endereço das informações dentro de uma tabela. Com as coordenadas/endereço de cada informações na uma tabela, conseguimos utilizar mecanismos de busca para filtrar subconjuntos (valor isolado, linha, coluna, matriz) de interesse que estão contidos na tabela. Sendo assim, os índices servem para especificar o endereço das informações dentro de uma matriz. Existem dois tipos de índice: o padrão (número das linhas e colunas - começa contar em 0), e o índice semântico (valores/labels de uma coluna definida e os rótulos das colunas). Um ponto negativo do índice padrão é que, em casos de reordenação das linhas/colunas, um mesmo valor pode mudar de índice (endereço). Para fazer as consultas por meio dos índices utilizamos as funções loc - para índices semânticos/labes e iloc - para índices padrões/números.

**AULA 3 Mensagem-chave: Qualidade de dados**: Os aspectos que influenciam na qualidade dos dados são: fonte, forma/dispositivo utilizado para coleta dos dados, período da coleta dos dados, balanceamento, dados faltantes, inconsistentes, redundantes, ruidosos, atípicos, tendenciosos. No caso de dados faltantes é importante identificarmos o mecanismos dessa ausência, ou seja, entender a origem ou motivo dos dados faltantes, que podem ser totalmente aleatórios - MCAR, parcilamente aleatórios - MAR (influenciado por outro atributo) ou não aleatório - MNAR (segue um padrão, é previsível). Para lidar com valores ausentes (mitigar os impactos negativos) as estratégias são: ignorar valores ausentes, excluir objetos (linhas) ou atributos (colunas), substituir por estimativas (média, mediana, moda) ou valores fictícios. É importante inserir na base de dados uma coluna com valores binários para identificar valores que foram incluídos ou tratados na base. Ter uma atenção a qualidade dos dados é importante para assegurar bons resultados em etapas seguintes de análise e modelagem.

**AULA 4**

*   **Transformação de dados**: Consiste em mudar o valor do dado para adaptar a informação para uma necessidade de análise ou modelagem. O processo de tranformação de dados vai depender da necessidade da transformação, que pode ser: anonimização, conversão normalização/padronização.

*   **Anonimização**: modificar o valor do dado com a finalidade de remover qualquer possibilidade de identificação direta ou indireta (engenharia reversa) de informações pessoais ou institucionais que possam constranger ou expor o indivíduo. No entanto, é importante que, no processo de anonimização, a essência e as características da informação original sejam mantidas sem a inserção de nenhum viés artificial. A aplicação do processo de anonimização vai depender do tipo do dado. Se for categórico, altera-se as classes por classes fictícias (evitar classes totalmente numéricas). Se for um dado ordinal, substituir por um valor numérico proporcional a posição da classe. Já se for uma valor numérica a anonimização deve ser realizada por meio de uma função matemática. No caso de valores inteiros, procurar manter o padrão numérico, ou seja, evitar divisões que gerem valores decimais e cuidar com subtrações que resultam em números inteiros. Por fim, é importante guardar as regras de anonimização, para o caso de precisar da traduzação de volta para os valores originais.

*   **Conversão**: Consistem em converter o tipo do dado, ou seja, entre qualitativo e quantitativo. A conversão de dados é necessária para viabilizar a implementação de modelos de predição ou classificação, por exemplos, que exigem um tipo específico de atributo para seu funcionamento. A conversão de valores depende do tipo do atributo. Se for de ordinal para numérico, basta substituir pela posição ordenada das classes ou qualquer valor proporcional. Se for um dado categórico para numérico, aplica-se a técnica oneHot que criar n colunas binárias, sendo n a quantiadade de classes desse atributo categórico. Já se for um dado numérico para qualitativo, divide o conjunto de dados em classes e nomeia-se cada uma das classes, fazendo a conversão de acordo com as classes criadas dentro do intervalo original.





**AULA 5: TRANSFORMAÇÃO DE DADOS**: consiste em técnicas que alteram os valores dos dados, porém, o tipo permanece o mesmo. Esses tipos de transformações podem auxiliar na visualização dos dados e na padronização das escalas e da distribuição. As transformações de dados são implementadas por diferentes estratégias:
* **Funções matemáticas**: são transformações realizas por funções matemáticas simples, como log(x) (para corrigir distribuições fortemente assimétricas), raiz(x) (para corrigir distribuições levemente assimétricas) e inversa 1/x (para dar ênfase em valores menores).

* **Padronização de escala**: é um tipo de transformação que ajusta os extremos da escala dos dados. Entre os principais métodos, destacam-se: minmax ou max. Uma limitação importante destas técnicas é quando o conjunto de dados têm outliers. Nesse caso, pode ser aplicando uma transformação por função matemática antes da padronização por escala.
* **Normalização**: são técnicas de padronização que ajustam a distribuição da escala e convertem a média para 0 e desvio padrão para 1. O método mais utilizado é o Z-score. Porém, caso o conjunto tenha outliers, o Z-score deve ser implementado com base em mediana e IQR (RobustScale).


Essas padronização são realizadas para ajustar os dados aos pré-requisidos dos modelos a serem implementados na fase de análise. Caso necessário, as transformação precisam ser revertidas para apresentação dos resultados.




**AULA 6 e 7: VISUALIZAÇÃO ESTATÍSTICA DE DADOS**: gráficos são recursos visuais utilizar para transmitir informações de forma rápida. Independente do tipo de visualização gráfica utilizada, alguns elementos são essenciais para representar adequadamente a informação: título (precisa, necessariamente, responder as perguntas o que? onde? quando?), legenda (se necessário), fonte dos dados, descrição dos eixos. Alguns exemplos de boas práticas na visualização estatística são: cuidado na escolha das cores, cuidado com as escalas, cuidado na escolha adequada do gráfico (que depende do tipo do dado que temos e da informação que gostariamos de transmitir), destaçar títulos e eixos no gráfico. Já alguns exemplos de más práticas são: uso de gráficos 3D (sem necessidade), cortar escala dos eixos, usar gráficos inadequados para o tipo de dado.

* **Gráfico de barras/colunas**: utilizado para finalidade de comparar valores entre classes ou categorias.
* **Gráfico de setor**: mostrar proporções entre classes. As proporções de cada classe devem somar 100%.
* **Gráfico de dispersão**: tem a finalidade de identicar padrões de associação entre variáveis numéricas e também agrupamentos entre classes.
* **Gráfico de linhas**: tem a finalidade de ilustrar padrões de tendência e sazonalidade temporal.
* **Gráfico de radar**: utilizado para comparar perfis de entidades (categorias) em função de atributos numpéricos. Evidencia pontos positivos e negativos entre as categorias. É importante que as variáveis numéricas estejam normalizadas na mesma escala.
* **Coordenadas Paralelas**: utilizado quando o objetivo é demonstrar o comportamento de uma entidade (categoria) em relação a múltiplas variáveis numéricas (normalizadas na mesma escala). O gráfico também pode ser utilizado para ilistrar associação de variáveis, bem como agrupamentos das entidades.
* **Histograma**: conhecer o formato da distribuição dos dados de uma variável numérica.
* **Box Plot (gráfico de caixa)**: utilizado quando o objetivo é conhecer um resumo estatístico de um conjunto de dados numéricos (medida de posição (mediana - onde os dadso estão concentrados), medida de dispersão (IQR - quanto os dados estão espalhados), quartis, amplitude, outliers, assimetria).
* **Gráfico de Violino**: utilizado para conhecer um resumo estatístico de um conjunto de dados numéricos, incluindo a sua distribuição de frequências.
* **Nuvem de Palavras**: tem a finalidade de explorar a existência de algum padrão em um conjunto de dados textual.
* **Mapa de calor**: é uma forma visual de ilustrar a associação ou padrão entre variáveis.

**AULA 8: ASSOCIAÇÃO E CORRELAÇÃO**: associação é um conceito que define a relação entre duas variáveis. Relação entre variáveis significa que, conhecendo o valor de uma, conseguimos estimar o valor da outra. Correlação é uma forma de quantificar, ou seja, resumir essa associação em um número. O coeficiente de correlação é um número, que varia entre -1 e 1, e que quantifica a força e a direção da associação. Uma associação pode ser linear ou monotônica, positivas ou negativas. Toda relação linear e monotônica, porém, nem toda relação monotônica é linear. É possivel também que duas variáveis não estejam associadas. Para associação linear, o coeficiente mais indicado é o coeficiente de Pearson. Para associações monotônicas o coeficiente mais indicado é o coeficiente de Spearman. Sempre que os coeficientes de Pearson e Spearman distoarem muito, é importante entender o motivo desta diferença, que pode ser, curvatura da associação, outliers, distribuições assimétricas. O uso simultâneo do gráfico de dispersão auxilia na investigação e entendimento dessas causas. Remoção de outliers e transformações das escalas podem auxiliar na análise de sensibilidade dos coeficientes aos valores extremos.

```python
# 1. Carragar DataFrame
dados = {
    'ID Vendedor': [101, 102, 101, 103, 104],
    'Região': ['Sul', 'NORTE', 'Sul', 'sul', 'Leste'],
    'FATURAMENTO': [5000, 3000, 5000, np.nan, 4500],
    'Email': [
        'ana@furb.br',
        'bruno@furb.br',
        'ana@furb.br',
        'carla@furb.br',
        'joao@furb.br',
    ]
}

df = pd.DataFrame(dados)
print(df)

# Carregar diretamento do Excel
#df = pd.read_excel("file_name")
```

```python
# 2. Renomeação das colunas

df1 = df.rename(columns={
    'ID Vendedor': 'ID_VENDEDOR',
    'Região': 'REGIAO',
    'Email': 'EMAIL'
})

print(df1)
```

```python
# 3. Eliminação de duplicatas

df2 = df1.drop_duplicates(subset=['ID_VENDEDOR'])
print(df2)
```

```python
# 4. Tratamento de Dados Ausentes

df2['FATURAMENTO'] = df2['FATURAMENTO'].fillna(df2['FATURAMENTO'].mean())

print(df2)
```

```python
# 5. Criação de uma coluna calculada

df2['FATURAMENTO-ANUAL'] = df2['FATURAMENTO']*12
print(df2)
```

```python
# 5. Criação de uma nova coluna com lista de valores

df2['PERC-COMISSAO'] = [10, 20, 15, 10]
print(df2)
```

```python
# 5. Criação de uma nova coluna operando com mais de uma coluna

df2['LUCRO'] = df2['FATURAMENTO-ANUAL'] - df2['PERC-COMISSAO']/100 * df2['FATURAMENTO-ANUAL']
print(df2)
```

```python
# 6. Padronização e ordenação

df2['REGIAO'] = df2['REGIAO'].str.lower()
df2 = df2.sort_values(by='FATURAMENTO', ascending=False)

print(df2)
```

```python
# 7. Exportação Final

df2.to_excel('dados_processados.xlsx', index=False)
df2.to_csv('dados_processados.csv', index=False)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'nome': ['Ana', 'Bruno', 'Carlos', 'Daniel', 'Ana', 'Elena', 'Fabio', 'Gisele'],
    'idade': [21, 176, 1, 26, 21, 41, 34, 28],
    'data_nascimento': ['1999-05-10', '1844-01-01', '2019-03-20', '1994-08-15', '1999-05-10', '1979-11-30', '1986-02-12', '1992-07-05'],
    'temperatura': [36.5, 37.0, -5.0, 36.8, 36.5, 41.5, 37.2, 36.6],
    'frequencia_cardiaca': [75, 80, 72, np.nan, 75, 110, 999, 82],
    'escolaridade': ['Graduação', 'Mestrado', 'Doutorado', 'Ensino Médio', 'Graduação', 'Graduação', 'Ensino Médio', 'Pós-graduação']
})

print(df)
```

```python
# Identifica e remove registros idênticos
print(df[df.duplicated(keep=False)])
df2 = df.drop_duplicates(keep='last')
print('\n', df2)
```

```python
df2 = df2.drop(columns=['data_nascimento'])
print(df2)
```

```python
# Tratamento de dados ausentes
# Contar quantos nulos existem por coluna
print("\nDiagnóstico de NaNs:\n", df2.isnull().sum(), '\n')
```

```python
# Tratamento de dados inconsistentes e outliers
# Criando o indicador de ausência
frequencia_ausente = df2['frequencia_cardiaca'].isna().astype(int)
df2.insert(loc=df2.columns.get_loc("frequencia_cardiaca") + 1,
          column='FREQ_CARDIACA_AUS',
          value=frequencia_ausente)

# Preencher a frequência cardíaca vazia com a média
mediana_frequencia = round(df2['frequencia_cardiaca'].median())
df2['frequencia_cardiaca'] = df2['frequencia_cardiaca'].fillna(mediana_frequencia)

print(df2)
```

```python
df3 = df2.copy()
```

```python
# CORREÇÕES DE INCONSISTÊNCIAS E RUÍDOS

# Criando o indicador de dado corrigido
idade_corrigida = (df3["idade"] > 120).astype(int)
df3.insert(loc=df3.columns.get_loc("idade") + 1,
          column='IDADE_COR',
          value=idade_corrigida)

# Identificar e tratar idade impossível
media_idade = round(df3.loc[df3['idade'] <= 120, 'idade'].mean())
df3.loc[df3['idade'] > 120, 'idade'] = media_idade

# Criando o indicador de dado corrigido
temp_corrigida = (df3["temperatura"] < 30).astype(int)
df3.insert(loc=df3.columns.get_loc("temperatura") + 1,
          column='TEMP_COR',
          value=temp_corrigida)

# Identificar e tratar temperatura impossível (ex: < 30°C)
media_temp = round(df3.loc[df3['temperatura'] >= 30, 'temperatura'].mean(), 1)
df3.loc[df3['temperatura'] < 30, 'temperatura'] = media_temp


# Identificar e tratar ruído na frequência cardíaca
media_frequencia = round(df3.loc[df['frequencia_cardiaca'] <= 300, 'frequencia_cardiaca'].mean())
df3.loc[df['frequencia_cardiaca'] == 999, 'frequencia_cardiaca'] = media_frequencia

print(df3)
```

```python
# Importando a função de transformação ordinal do scikit-learn
from sklearn.preprocessing import OrdinalEncoder

ordem = ['Primeiro', 'Segundo', 'Terceiro', 'Quarto', 'Quinto']
codificador = OrdinalEncoder(categories=[ordem])  # atribui valores numéricos baseados na ordem da lista
df['Clas_convertida'] = codificador.fit_transform(df[['Classificação']])

print(df)
```

```python
######################################
## CONVERSÃO: NOMINAL PARA NUMÉRICO ##
######################################

from sklearn.preprocessing import OneHotEncoder

# criar uma coluna binária para cada categoria única encontrada
encoder = OneHotEncoder()

# transformar a coluna categórica em uma matriz binária (Dummificação)
estados = encoder.fit_transform(df[['Estado']]).toarray()

# converte a matriz de números binários em um DataFrame do Pandas
estados_df = pd.DataFrame(estados, columns=encoder.categories_[0])

# Concatenando as novas colunas e removendo a coluna original
df = pd.concat([df, estados_df], axis=1)
df.drop('Estado', axis=1, inplace=True)

print(df)
```

```python
import pandas as pd
import numpy as np

dados = {
    'Rendimento_Anual': [12000, 45000, 180000, 2500000],
    'Erros_Producao': [0, 4, 9, 25],
    'Tempo_Conclusao_Min': [2, 5, 10, 20],
    'Variacao_Temperatura': [-5.2, 3.1, -1.8, 0.4]
}

df = pd.DataFrame(dados)

# 1. Logarítmica
df['Rendimento_Log'] = np.log(df['Rendimento_Anual'])

# 2. Raiz Quadrada
df['Erros_Raiz'] = np.sqrt(df['Erros_Producao'])

# 3. Recíproco
df['Taxa_Produtividade'] = 1 / df['Tempo_Conclusao_Min']

# 4. Módulo
df['Variacao_Absoluta'] = df['Variacao_Temperatura'].abs()

print(df)
```

**Normalização**

```python
import pandas as pd

dados = {
    'Candidato': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Anos_Experiencia': [1, 3, 5, 8, 10],
    'Pontuacao_Teste': [400, 550, 700, 850, 1000]
}
df = pd.DataFrame(dados)
df.head()
```

```python
# Aplicar as transformações na coluna 'Pontuacao_Teste'
x = df['Pontuacao_Teste']
min_x = x.min()
max_x = x.max()
media_x = x.mean()
desvio_x = x.std()

# 1. Normalização pelo Máximo
df['Teste_Max'] = x / max_x

# 2. Normalização Min-Max ([0, 1])
df['Teste_MinMax'] = (x - min_x) / (max_x - min_x)

# 3. Escalonamento em Intervalo Personalizado [a, b]
a, b = -2, 2
df['Teste_Intervalo'] = a + ((x - min_x) * (b - a)) / (max_x - min_x)

# 4. Padronização Z-Score (Média = 0, Desvio Padrão = 1)
df['Teste_ZScore'] = (x - media_x) / desvio_x


print(df[['Candidato', 'Pontuacao_Teste', 'Teste_Max', 'Teste_MinMax',
          'Teste_Intervalo', 'Teste_ZScore']].round(3))
```

## Gráfico de Radar

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento do dataset
df_brasil = pd.read_csv('BRAZIL_CITIES.csv', sep=';')

# Cria a coluna 'REGION' no DataFrame, mapeando cada UF (estado) para sua região
mapa_regioes = {
    'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte',
    'RR': 'Norte', 'TO': 'Norte', 'AL': 'Nordeste', 'BA': 'Nordeste',
    'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
    'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste', 'DF': 'Centro-Oeste',
    'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df_brasil['REGION'] = df_brasil['STATE'].map(mapa_regioes)

# Seleciona as variáveis que vamos comparar no radar:
variaveis_idhm = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao']
variaveis_gva = ['GVA_AGROPEC', 'GVA_INDUSTRY', 'GVA_SERVICES', 'GVA_PUBLIC']

# Calcula a média de cada variável, agrupando os municípios por região
dados = df_brasil.groupby('REGION')[variaveis_idhm + variaveis_gva].mean()

# O IDHM está em escala 0-1, enquanto o GVA será convertido para %.
# Multiplicar por 100 coloca o IDHM na mesma ordem de grandeza das demais
# variáveis, para nenhum eixo "dominar" o gráfico por causa da escala.
dados[variaveis_idhm] = dados[variaveis_idhm] * 100

# O GVA vem em valores monetários brutos, que variam muito entre regiões
# Por isso, convertemos cada setor para sua % de participação dentro da própria região.
dados['GVA_TOTAL'] = dados[variaveis_gva].sum(axis=1)
for col in variaveis_gva:
    dados[col] = (dados[col] / dados['GVA_TOTAL']) * 100
dados = dados.drop(columns='GVA_TOTAL')  # coluna auxiliar não é mais necessária

# Seleciona apenas 3 regiões para não poluir o gráfico com muitos polígonos
dados = dados.loc[['Sul', 'Nordeste', 'Sudeste']]

# Define os ângulos dos eixos do radar
categorias = list(dados.columns)
n_vars = len(categorias)   # quantidade de eixos
# criar sequência de número igualmente espaçados entre 0 e 2pi
angulos = np.linspace(0, 2*np.pi, n_vars, endpoint=False).tolist()
angulos += angulos[:1]  # repete o primeiro ângulo no final para fechar o polígono

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True)) #definir malhar polar
cores = {'Sul': '#5B8FF9', 'Nordeste': '#F98C5C', 'Sudeste': '#5AD8A6'}

# Desenha um polígono por região
for regiao in dados.index:
    valores = dados.loc[regiao].tolist()
    valores += valores[:1]
    ax.plot(angulos, valores, linewidth=2, label=regiao, color=cores[regiao])
    ax.fill(angulos, valores, alpha=0.1, color=cores[regiao])

# Configura os rótulos de cada eixo (nome das variáveis ao redor do círculo)
ax.set_xticks(angulos[:-1])
ax.set_xticklabels(categorias, fontsize=9)
ax.set_yticklabels([])  # oculta os valores numéricos do raio
ax.set_title('Perfil comparativo entre regiões: IDH e composição econômica (%)',
             fontsize=12, pad=20)
ax.legend(loc='best')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

# Carregamento do dataset
df_brasil = pd.read_csv('BRAZIL_CITIES.csv', sep=';')

# Cria a coluna 'REGION' no DataFrame
mapa_regioes = {
    'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte',
    'RR': 'Norte', 'TO': 'Norte', 'AL': 'Nordeste', 'BA': 'Nordeste',
    'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
    'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste', 'DF': 'Centro-Oeste',
    'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df_brasil['REGION'] = df_brasil['STATE'].map(mapa_regioes)

# Seleciona só as colunas que vamos usar e remove municípios com algum valor faltando
colunas = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao', 'ESTIMATED_POP', 'GDP', 'REGION']
# sample(100) pega 100 municípios aleatórios, para o gráfico não ficar poluído
amostra = df_brasil[colunas].dropna().sample(100, random_state=1)

# Normalização min-max (0 a 1) de cada variável quantitativa.
variaveis_num = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao', 'ESTIMATED_POP', 'GDP']
amostra_norm = amostra.copy()
for col in variaveis_num:
    amostra_norm[col] = (amostra[col] - amostra[col].min()) / (amostra[col].max() - amostra[col].min())

# Construção do gráfico de coordenadas paralelas.
# A função parallel_coordinates já vem pronta no pandas - não é preciso desenhar
# os eixos manualmente (diferente do gráfico de radar, feito com matplotlib puro).
fig, ax = plt.subplots(figsize=(11, 6))
parallel_coordinates(
    amostra_norm,
    class_column='REGION',   # coluna categórica usada para colorir cada linha
    cols=variaveis_num,      # variáveis que formam os eixos verticais, na ordem desejada
    ax=ax,
    alpha=0.5                # transparência nas linhas, para reduzir a sobreposição visual
)

# Posiciona a legenda um pouco para fora da área do gráfico, à direita
ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1))

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carrega o dataset de cidades brasileiras
df = pd.read_csv('BRAZIL_CITIES.csv', sep=';')
```

## Tabelas de contingência

```python
# ------------------------------------------------------------
# 1. Criar a variável REGIAO
# ------------------------------------------------------------
regioes = {
    'AC': 'Norte', 'AP': 'Norte', 'AM': 'Norte', 'PA': 'Norte',
    'RO': 'Norte', 'RR': 'Norte', 'TO': 'Norte',
    'AL': 'Nordeste', 'BA': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste',
    'PB': 'Nordeste', 'PE': 'Nordeste', 'PI': 'Nordeste', 'RN': 'Nordeste',
    'SE': 'Nordeste',
    'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'MS': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df['REGIAO'] = df['STATE'].map(regioes)

# ------------------------------------------------------------
# 2. Discretizar o IDHM nas faixas do PNUD
# ------------------------------------------------------------
limites = [0, 0.5, 0.6, 0.7, 0.8, 1]
faixas = ['Muito baixo', 'Baixo', 'Médio', 'Alto', 'Muito alto']
df['FAIXA_IDHM'] = pd.cut(df['IDHM'], bins=limites, labels=faixas, right=False)

# Verificar valores ausentes antes de cruzar (o crosstab os ignora)
print('Municípios sem faixa de IDHM:', df['FAIXA_IDHM'].isna().sum())
```

```python
# ============================================================
# EXEMPLO 1: Região x faixa de IDHM
# ============================================================

# ------------------------------------------------------------
# 1. Frequências absolutas, com totais marginais
# ------------------------------------------------------------
tabela = pd.crosstab(df['REGIAO'], df['FAIXA_IDHM'],
                     margins=True, margins_name='Total')
print('\n--- Frequências absolutas ---')
print(tabela)

# ------------------------------------------------------------
# 2. As três formas de calcular proporções (em %)
# ------------------------------------------------------------
# Pelo total geral: cada célula dividida pelo total da tabela
prop_total = pd.crosstab(df['REGIAO'], df['FAIXA_IDHM'], normalize='all')*100

# Por linha: cada célula dividida pelo total da sua região
prop_linha = pd.crosstab(df['REGIAO'], df['FAIXA_IDHM'], normalize='index')*100

# Por coluna: cada célula dividida pelo total da sua faixa de IDHM
prop_coluna = pd.crosstab(df['REGIAO'],df['FAIXA_IDHM'],normalize='columns')*100

print('\n--- % pelo total geral ---')
print(prop_total.round(1))
print('\n--- % por linha (dentro de cada região) ---')
print(prop_linha.round(1))
print('\n--- % por coluna (dentro de cada faixa de IDHM) ---')
print(prop_coluna.round(1))
```

```python
# ============================================================
# EXEMPLO 2: Região x tipo rural-urbano
# ============================================================

# ------------------------------------------------------------
# 1. Remover a categoria "Sem classificação" e cruzar
# ------------------------------------------------------------
df_ru = df[df['RURAL_URBAN'] != 'Sem classificação']

ru_linha = pd.crosstab(df_ru['REGIAO'], df_ru['RURAL_URBAN'], normalize='index') * 100
ru_coluna = pd.crosstab(df_ru['REGIAO'], df_ru['RURAL_URBAN'], normalize='columns') * 100

print('\n--- Região x tipo rural-urbano: % por linha ---')
print(ru_linha.round(1))
print('\n--- Região x tipo rural-urbano: % por coluna ---')
print(ru_coluna.round(1))
```

## Associação Categórica

```python
from scipy.stats.contingency import association

# ------------------------------------------------------------
# 1. Criar as variáveis categóricas
# ------------------------------------------------------------
regioes = {
    'AC': 'Norte', 'AP': 'Norte', 'AM': 'Norte', 'PA': 'Norte',
    'RO': 'Norte', 'RR': 'Norte', 'TO': 'Norte',
    'AL': 'Nordeste', 'BA': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste',
    'PB': 'Nordeste', 'PE': 'Nordeste', 'PI': 'Nordeste', 'RN': 'Nordeste',
    'SE': 'Nordeste',
    'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'MS': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df['REGIAO'] = df['STATE'].map(regioes)

# Faixa de IDHM (faixas do PNUD)
limites = [0, 0.5, 0.6, 0.7, 0.8, 1]
faixas = ['Muito baixo', 'Baixo', 'Médio', 'Alto', 'Muito alto']
df['FAIXA_IDHM'] = pd.cut(df['IDHM'], bins=limites, labels=faixas, right=False)

# Presença de Uber: a coluna UBER tem 1 onde há Uber e fica vazia nos demais
df['TEM_UBER'] = df['UBER'].notna().map({True: 'Sim', False: 'Não'})

# Tipo rural-urbano: remover a categoria "Sem classificação"
df_ru = df[df['RURAL_URBAN'] != 'Sem classificação']

# ------------------------------------------------------------
# 2. Montar as tabelas de contingência (frequências absolutas)
# ------------------------------------------------------------
tabela_idhm = pd.crosstab(df['REGIAO'], df['FAIXA_IDHM'])
tabela_ru = pd.crosstab(df_ru['REGIAO'], df_ru['RURAL_URBAN'])
tabela_uber = pd.crosstab(df['REGIAO'], df['TEM_UBER'])

# ------------------------------------------------------------
# 3. Calcular o V de Cramér de cada tabela
# ------------------------------------------------------------
v_idhm = association(tabela_idhm, method='cramer')
v_ru = association(tabela_ru, method='cramer')
v_uber = association(tabela_uber, method='cramer')

print('V de Cramér')
print(f'Região x faixa de IDHM      : {v_idhm:.2f}')
print(f'Região x tipo rural-urbano  : {v_ru:.2f}')
print(f'Região x presença de Uber   : {v_uber:.2f}')
```
