**MENSAGEM CHAVE DA AULA 2**:


*   REESTRUTURAÇÃO E PIVOTAMENTO DE DADOS: pivotamento consiste em compactar uma tabela no padrão de máquina (longa) para o padrão humano (largo). Podemos fazer isso utilizando a função do Pandas pivot(). Já a reestruturação de uma tabela consiste em transformar uma tabela do formato humano (largo) para o formato de máquina (longo), ou seja, formato organizado. Podemos fazer a reestruturação utilizando a função do Pandas: melt().


*   CONCATENAÇÃO E FUSÃO DE CONJUNTO DE DADOS:consiste no conceito de unir tabelas de diferentes formas. A concatenação consiste no empilhamento de tabelas, desde que tenham as mesmas informações (colunas) ou cenários (linhas). O empilhamento pode acontecer na vertical (aumenta o n. de linhas, axis=0) ou na horinzontal (aumenta o n. de colunas, axis=1). A concatenação é implementada por meio da função pd.concat(). Já a fusão consiste em combinar duas tabelas a partir de uma coluna comum (chave). Existe mais de uma forma de concatenação, por exemplo: left (tudo que tem na primeira), right (tudo o que tem na segunda), inner (tem nas duas simultaneamente), outer (soma das duas tabelas), entre outras. A fusão de tabelas e realizada por meio da função pd.merge(), da biblioteca Pandas.

*   INDEXAÇÃO E FERRAMENTAS DE BUSCA: a indezação serve como coordenada para filtragem de dados com base no endereço de linhas e colunas. A ordem de leitura é sempre primeiro linha e depois coluna. Os endereços/coordenadas das linhas e colunas pode ser o numérico padrão (índice padrão) ou alguma coluna especificada (indice semantico). Um ponto de atenção é que o índice padrão não é fixo da linha, caso a tabela seja reordenada. Nesses casos de reordenação dos dados, o índice semântico será mais assertivo. As principais ferramentas de busca baseadas en índice são: iloc() - baseado em número, ou seja, índice padrão, e a função loc() - baseada no índice semântico. Além disso, a soma de dataframes também ocorre em função dos índices e não por célula correspondente.

**MENSAGEM CHAVE DA AULA 3**:


*   QUALIDADE DE DADOS: a qualidade dos dados é importante para garantir/assegurar bons resultados na análise ou modelagem. A qualidade de um conjunto de dados é vericada com base na existência de valores ausentes, inconsistentes, redundates, ruidosos, tendenciosos, como também em função da forma como os dados foram obtidos (fonte, equipamento e dispositivos utilizados, período de tempo, ...).

* TRATAMENTO DE VALORES AUSENTES: é importante entender os mecanismos que explicam o motivo da ausência daquelas informações (dados), ou seja, se aquela ausência é aleatória (não possui nenhum padrão e não é previsível), se ela depende da informação de outros atributos, ou se ela é previsível (não aleatória) e tem algum padrão que justifique a sua ausência. As formas de lidar com os valores ausentes são: desconsiderar a ausência (caso futuros modelos suportem essa condição); descartar os objetos (linhas), descartar os atributos (colunas); substituir por um valor fictício representativo; ou substituir por uma estimativa. Além disso, é muito importante, criar uma nova coluna preditiva com valores binários indicando a ausência (1) ou não ausência (0) de valor no atributo original.

**MENSAGEM CHAVE DA AULA 4**:


*   TRANSFORMAÇÃO DE DADOS: Consiste em um conjunto de regras para alterar as informações de atributos. A finalidade da transformação de dados consiste em alterar os valores para garantir a anonimização da informação, bem como, converter e transformar valores e escalas dentro dos padrões necessários para a análise. Tudo isso, sem descaracterizar a essencial informação (variabilidade e distribuição).

* ANONIMIZAÇÃO: consiste em mascarar os valores para manter a privacidade e segurança de dados pessoais ou institucionais. A anonimização é realizada em função do tipo do dado do atributo (categórico, ordinal e numérico). A anonimização dos atributos categóricos é feita a partir da substituição das classes por novos valores fictícios (ex: letras). Já os atributos ordinais, pode ser substituido por uma sequência numérica que respeite a ordem e a sequência das classes do atributo original (ex: 1, 2, 3; ou 10, 20, 30). Por fim, os atributos numpericos podem ser anonimizados através de funções matemáticas, normalmente, envolvendo somas, subtrações e multiplicação. As respectivas funções inversas podem ser utilizadas para reversão dos valores (recuperar valores originais). É importante que o analista crie uma tabela com as regras de anonimização utilizadas para possibilitar a conversão de volta para os valores originais, caso necessário.



**MENSAGEM CHAVE DA AULA 5**:


*   CONVERSÃO DE DADOS: transformar/mudar o valor é a categoria do dados, ou seja, de quantitativo para qualitativo ou vice-versa. a covnersão de valores é realizada para ajustar/adaptar os dados aos requisitos de funcionamento dos modelos e métodos de análise utilizados em etapas seguintes do projeto. A implementação da conversão de dados depende do tipo da conversão que precisa ser realizada. (a) Para converter um dado ordinal em numérico, basta organizar os atributos em ordem e substituir pela sua respectiva posição. (b) Já para converter um dado nominal em numérico, precisamos realizar uma conversão do tipo binária (1-de-n, one-hot, hot-encoding), onde, cada categoria do atributo nominal vira uma nova coluna binária. Essa estratégia evita introduzis nos dados convertidos informações de ordem e distância que não existem nos dados originais. Quando um atributo nominal possui muitas categorias distintas, uma estratégia é utilizar combinações de pseudo-atributos numéricos. (C) Por fim, para converter uma atributo numérico para qualitativo, precisamos discretizar o intervalo original em subintervalos (utilizando critérios técnicos ou estatísticos), nomear cada um deles, e classificar cada dado numérico de acordo com seu respectivo intervalo de pertencimento.

**MENSAGEM CHAVE DA AULA 6**:


*   TRANSFORMAÇÃO POR NORMALIZAÇÃO: são transformações dos valores de um atributo para ajustar sua escala, para normalizar seus dados (ajustar sua distribuição para o padrão normal) ou para melhorar a visualização. Esse tipo de transformação pode ser realizado: (a) por meio de funções matemática simples (logaritmo, raiz, inverso), que tem a finalidade de corrigir a assimetria da distribuição dos dados; ou (b) por funções/equações matemáticas de normalização (escala e distribuição), como por exemplo, a função max (ajusta a escala da variável para que o valor máximo seja 1 e os demais valores proporcionalmente menores), minmax (ajusta o inicio e final da escala), z-score (transforma a escala, de modo que a média do conjunto de dados seja 0 e o desvio padrão 1).

*   VISUALIZAÇÃO ESTATÍSTICA DE DADOS: a visualização de dados é essencial, pois, facilita o entendimento e identificação de informações de grandes bases de dados, de forma muito mais rápida do que a exposição por meio de tabelas. Uma gráfico, independente do seu tipo, deve apresentar alguns elementos que são indispensáveis para sua compreensão, como exemplo: título (responde o que? onde? quando?), legenda (quando necessário), fonte dos dados, título dos eixos.
**Exemplos de maus usos de um gráfico**: gráfico 3D, não truncar a escala dos eixos (sempre começar em 0), cuidar com a escolha das cores.

* **Gráfico de barras**: utilizado para comparação de categorias, principalmente, para evidenciar diferenças.
* **Gráfico de Setor (pizza)**: tem a finalidade de comparar proporções de categorias, com o objetivo de evidenciar categorias em destaque. Evitar utilizar muitas categorias (máximo 6) ou categorias que tenham proporção muito semelhante entre si.
* **Gráfico de linha**: utilizado para ilustrar a evolução de alguma variável no tempo ou em alguma grandeza que tenha algum tipo de ordem natural. Quando utilizados dados discretos em uma gráfico de linhas é recomentado adotar uma linha pontilhada para não passar a falsa impressão de dados contínuos.
* **Gráfico de dispersão**: é utilizado para ilustrar a associação/correlação entre variáveis numéricas e também para identificar agrupamentos.

**MENSAGEM CHAVE DA AULA 7**:


*   **Gráfico de Radar**: tem a finalidade de comparar o perfil de diferentes entidades (classes de uma variável nominal) e destacar pontos fortes e fracos de cada perfil. Os eixos de um gráfico de radar são variáveis numéricas, cujas escalas precisam, necessariamente, estar normalizadas. Não utilizar mais que 5 ou 6 entidades em um mesmo gráfico.

*   **Gráfico de Coordenadas Paralelas**: utilizado para comparar o perfil de até várias entidades simultaneamente, também para identificar agrupamentos entre as entidades (linhas que seguem um percurso semelhante entre os eixos), e tambem para mostrar associação entre variáveis vizinhas.
*   **Histograma**: visualizar a distribuição de frequência dos dados. Ele só deve ser utilizado em variáveis contínuas ou discretas com muitos valores.
*   **Box Plot (Gráfico caixa)**: utilizado para finalidades mais técnicas de conhecer e visualizar um resumo estatístico descritivo de um conjunto de dados.
*   **Gráfico de violino**: (histograma + boxplot) utilizado quando, além de um resumo da estatística descritiva dos dados, precisamos conhecer o formato da distribuição de frequência do conjunto de dados.
*   **Nuvem de palavras**: é um gráfico exploratório para idenitificar padrões em um atributro textual.
*   **Mapa de calor**: visualizar a associação entre variáveis numericas (matriz correlação).

```python
# 1. Carragar DataFrame
dados = {
    'ID Vendedor': [101, 102, 101, 103, 104],
    'Região': ['Sul', 'NORTE', 'Sul', 'sul', 'Leste'],
    'FATURAMENTO': [5000, 3000, 5000, np.nan, 4500],
    'Email': [
        'ana@furb.br',
        'bruno@furb.br',
        'ana@furb.br',
        'carla@furb.br',
        'joao@furb.br',
    ]
}

df = pd.DataFrame(dados)
print(df)

# Carregar diretamento do Excel
#df = pd.read_excel("file_name")
```

```python
# 2. Renomeação das colunas

df1 = df.rename(columns={
    'ID Vendedor': 'ID_VENDEDOR',
    'Região': 'REGIAO',
    'Email': 'EMAIL'
})

print(df1)
```

```python
# 3. Eliminação de duplicatas

df2 = df1.drop_duplicates(subset=['EMAIL'])
print(df2)
```

```python
# 4. Tratamento de Dados Ausentes

df2['FATURAMENTO'] = df2['FATURAMENTO'].fillna(0)

print(df2)
```

```python
# 5. Criação de uma coluna calculada

df2['FATURAMENTO-ANUAL'] = df2['FATURAMENTO']*12
print(df2)
```

```python
# 5. Criação de uma nova coluna com lista de valores

df2['PERC-COMISSAO'] = [10, 20, 15, 10]
print(df2)
```

```python
# 5. Criação de uma nova coluna operando com mais de uma coluna

df2['LUCRO'] = df2['FATURAMENTO-ANUAL'] - df2['PERC-COMISSAO']/100 * df2['FATURAMENTO-ANUAL']
print(df2)
```

```python
# 6. Padronização e ordenação

df2['REGIAO'] = df2['REGIAO'].str.lower()
df2 = df2.sort_values(by='FATURAMENTO', ascending=False)

print(df2)
```

```python
# 7. Exportação Final

df2.to_excel('dados_processados.xlsx', index=False)
df2.to_csv('dados_processados.csv', index=False)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'nome': ['Ana', 'Bruno', 'Carlos', 'Daniel', 'Ana', 'Elena', 'Fabio', 'Gisele'],
    'idade': [21, 176, 1, 26, 21, 41, 34, 28],
    'data_nascimento': ['1999-05-10', '1844-01-01', '2019-03-20', '1994-08-15', '1999-05-10', '1979-11-30', '1986-02-12', '1992-07-05'],
    'temperatura': [36.5, 37.0, -5.0, 36.8, 36.5, 41.5, 37.2, 36.6],
    'frequencia_cardiaca': [75, 80, 72, np.nan, 75, 110, 999, 82],
    'escolaridade': ['Graduação', 'Mestrado', 'Doutorado', 'Ensino Médio', 'Graduação', 'Graduação', 'Ensino Médio', 'Pós-graduação']
})

df.head(10)
```

```python
df2 = df2.drop(columns=['data_nascimento'])
print(df2)
```

```python
# Tratamento de dados ausentes
# Contar quantos nulos existem por coluna
print("\nDiagnóstico de NaNs:\n", df2.isnull().sum(), '\n')
```

```python
# Tratamento de dados ausentes, inconsistentes e outliers
# Criando o indicador de ausência
frequencia_ausente = df2['frequencia_cardiaca'].isna().astype(int)
df2.insert(loc=df2.columns.get_loc("frequencia_cardiaca") + 1,
          column='FREQ_CARDIACA_AUS',
          value=frequencia_ausente)

# Preencher a frequência cardíaca vazia com a média
mediana_frequencia = round(df2['frequencia_cardiaca'].median())
df2['frequencia_cardiaca'] = df2['frequencia_cardiaca'].fillna(mediana_frequencia)

print(df2)
```

```python
df3 = df2
```

```python
# CORREÇÕES DE INCONSISTÊNCIAS E RUÍDOS

# Criando o indicador de dado corrigido
idade_corrigida = (df3['idade'] > 120).astype(int)
df3.insert(loc=df3.columns.get_loc("idade") + 1,
          column='IDADE_COR',
          value=idade_corrigida)

# Identificar e tratar idade impossível
media_idade = round(df3.loc[df3['idade'] <= 120, 'idade'].mean())
df3.loc[df3['idade'] > 120, 'idade'] = media_idade


# Identificar e tratar temperatura impossível (ex: < 30°C)
media_temp = round(df3.loc[df3['temperatura'] >= 30, 'temperatura'].mean(), 1)
df3.loc[df3['temperatura'] < 30, 'temperatura'] = media_temp


# Identificar e tratar ruído na frequência cardíaca
media_frequencia = round(df3.loc[df['frequencia_cardiaca'] <= 300, 'frequencia_cardiaca'].mean())
df3.loc[df['frequencia_cardiaca'] > 300, 'frequencia_cardiaca'] = media_frequencia

print(df3)
```

**Normalização**

```python
import pandas as pd

dados = {
    'Candidato': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Anos_Experiencia': [1, 3, 5, 8, 10],
    'Pontuacao_Teste': [400, 550, 700, 850, 1000]
}
df = pd.DataFrame(dados)
df.head()
```

```python
# Aplicar as transformações na coluna 'Pontuacao_Teste'
x = df['Pontuacao_Teste']
min_x = x.min()
max_x = x.max()
media_x = x.mean()
desvio_x = x.std()

# 1. Normalização pelo Máximo
df['Teste_Max'] = x / max_x

# 2. Normalização Min-Max ([0, 1])
df['Teste_MinMax'] = (x - min_x) / (max_x - min_x)

# 3. Escalonamento em Intervalo Personalizado [a, b]
a, b = -2, 2
df['Teste_Intervalo'] = a + ((x - min_x) * (b - a)) / (max_x - min_x)

# 4. Padronização Z-Score (Média = 0, Desvio Padrão = 1)
df['Teste_ZScore'] = (x - media_x) / desvio_x


print(df[['Candidato', 'Pontuacao_Teste', 'Teste_Max', 'Teste_MinMax',
          'Teste_Intervalo', 'Teste_ZScore']].round(3))
```

## Gráfico de Radar

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Carregamento do dataset
df_brasil = pd.read_csv('BRAZIL_CITIES.csv', sep=';')

# Cria a coluna 'REGION' no DataFrame, mapeando cada UF (estado) para sua região
mapa_regioes = {
    'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte',
    'RR': 'Norte', 'TO': 'Norte', 'AL': 'Nordeste', 'BA': 'Nordeste',
    'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
    'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste', 'DF': 'Centro-Oeste',
    'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df_brasil['REGION'] = df_brasil['STATE'].map(mapa_regioes)

# Seleciona as variáveis que vamos comparar no radar:
variaveis_idhm = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao']
variaveis_gva = ['GVA_AGROPEC', 'GVA_INDUSTRY', 'GVA_SERVICES', 'GVA_PUBLIC']

# Calcula a média de cada variável, agrupando os municípios por região
dados = df_brasil.groupby('REGION')[variaveis_idhm + variaveis_gva].mean()

# O IDHM está em escala 0-1, enquanto o GVA será convertido para %.
# Multiplicar por 100 coloca o IDHM na mesma ordem de grandeza das demais
# variáveis, para nenhum eixo "dominar" o gráfico por causa da escala.
dados[variaveis_idhm] = dados[variaveis_idhm] * 100

# O GVA vem em valores monetários brutos, que variam muito entre regiões
# Por isso, convertemos cada setor para sua % de participação dentro da própria região.
dados['GVA_TOTAL'] = dados[variaveis_gva].sum(axis=1)
for col in variaveis_gva:
    dados[col] = (dados[col] / dados['GVA_TOTAL']) * 100
dados = dados.drop(columns='GVA_TOTAL')  # coluna auxiliar não é mais necessária

# Seleciona apenas 3 regiões para não poluir o gráfico com muitos polígonos
dados = dados.loc[['Sul', 'Nordeste', 'Sudeste']]

# Define os ângulos dos eixos do radar
categorias = list(dados.columns)
n_vars = len(categorias)   # quantidade de eixos
# criar sequência de número igualmente espaçados entre 0 e 2pi
angulos = np.linspace(0, 2*np.pi, n_vars, endpoint=False).tolist()
angulos += angulos[:1]  # repete o primeiro ângulo no final para fechar o polígono

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True)) #definir malhar polar
cores = {'Sul': '#5B8FF9', 'Nordeste': '#F98C5C', 'Sudeste': '#5AD8A6'}

# Desenha um polígono por região
for regiao in dados.index:
    valores = dados.loc[regiao].tolist()
    valores += valores[:1]
    ax.plot(angulos, valores, linewidth=2, label=regiao, color=cores[regiao])
    ax.fill(angulos, valores, alpha=0.1, color=cores[regiao])

# Configura os rótulos de cada eixo (nome das variáveis ao redor do círculo)
ax.set_xticks(angulos[:-1])
ax.set_xticklabels(categorias, fontsize=9)
ax.set_yticklabels([])  # oculta os valores numéricos do raio
ax.set_title('Perfil comparativo entre regiões: IDH e composição econômica (%)',
             fontsize=12, pad=20)
ax.legend(loc='best')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

# Carregamento do dataset
df_brasil = pd.read_csv('BRAZIL_CITIES.csv', sep=';')

# Cria a coluna 'REGION' no DataFrame
mapa_regioes = {
    'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte',
    'RR': 'Norte', 'TO': 'Norte', 'AL': 'Nordeste', 'BA': 'Nordeste',
    'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
    'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste', 'DF': 'Centro-Oeste',
    'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df_brasil['REGION'] = df_brasil['STATE'].map(mapa_regioes)

# Seleciona só as colunas que vamos usar e remove municípios com algum valor faltando
colunas = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao', 'ESTIMATED_POP', 'GDP', 'REGION']
# sample(100) pega 100 municípios aleatórios, para o gráfico não ficar poluído
amostra = df_brasil[colunas].dropna().sample(100, random_state=1)

# Normalização min-max (0 a 1) de cada variável quantitativa.
variaveis_num = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao', 'ESTIMATED_POP', 'GDP']
amostra_norm = amostra.copy()
for col in variaveis_num:
    amostra_norm[col] = (amostra[col] - amostra[col].min()) / (amostra[col].max() - amostra[col].min())

# Construção do gráfico de coordenadas paralelas.
# A função parallel_coordinates já vem pronta no pandas - não é preciso desenhar
# os eixos manualmente (diferente do gráfico de radar, feito com matplotlib puro).
fig, ax = plt.subplots(figsize=(11, 6))
parallel_coordinates(
    amostra_norm,
    class_column='REGION',   # coluna categórica usada para colorir cada linha
    cols=variaveis_num,      # variáveis que formam os eixos verticais, na ordem desejada
    ax=ax,
    alpha=0.5                # transparência nas linhas, para reduzir a sobreposição visual
)

# Posiciona a legenda um pouco para fora da área do gráfico, à direita
ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1))

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr

# Carrega o dataset de cidades brasileiras
df = pd.read_csv('BRAZIL_CITIES.csv', sep=';')
```

## Exemplo 1

```python
# Dados do exemplo
df = pd.DataFrame({
    'experiencia': [1, 2, 3, 4, 5],
    'salario': [2, 3, 5, 14, 6]
})

# Com scipy
r, _ = pearsonr(df['experiencia'], df['salario'])
rho, _ = spearmanr(df['experiencia'], df['salario'])
print(f'scipy  -> Pearson = {r:.2f} | Spearman = {rho:.2f}')

# Com pandas
r_pd = df['experiencia'].corr(df['salario'], method='pearson')
rho_pd = df['experiencia'].corr(df['salario'], method='spearman')
print(f'pandas -> Pearson = {r_pd:.2f} | Spearman = {rho_pd:.2f}')
```
